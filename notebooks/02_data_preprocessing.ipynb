```python
from sklearn.preprocessing import LabelEncoder
```

```python
area_encoder = LabelEncoder()
item_encoder = LabelEncoder()
```

```python
df["Area"] = area_encoder.fit_transform(df["Area"])
df["Item"] = item_encoder.fit_transform(df["Item"])
```
