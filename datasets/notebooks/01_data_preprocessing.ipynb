```python
import pandas as pd

crop = pd.read_csv("../raw/Crop_recommendation.csv")
yield_data = pd.read_csv("../raw/yield.csv")
yield_df = pd.read_csv("../raw/yield_df.csv")
weather = pd.read_csv("../raw/state_weather_data_1997_2020.csv")
soil = pd.read_csv("../raw/state_soil_data.csv")
rainfall = pd.read_csv("../raw/rainfall.csv")
temperature = pd.read_csv("../raw/temp.csv")
pesticides = pd.read_csv("../raw/pesticides.csv")
fao = pd.read_csv("../raw/FAOSTAT_data_en_7-26-2026.csv")
```

```python
if "Unnamed: 0" in yield_df.columns:
    yield_df.drop(columns=["Unnamed: 0"], inplace=True)
rainfall["average_rain_fall_mm_per_year"] = rainfall["average_rain_fall_mm_per_year"].fillna(
    rainfall["average_rain_fall_mm_per_year"].median()
)
```

```python
#Rainfall
rainfall.info()
rainfall["average_rain_fall_mm_per_year"] = pd.to_numeric(
    rainfall["average_rain_fall_mm_per_year"],
    errors="coerce"
)
rainfall["average_rain_fall_mm_per_year"].fillna(
    rainfall["average_rain_fall_mm_per_year"].median(),
    inplace=True
)
```

```python
if "Unnamed: 0" in yield_df.columns:
    yield_df.drop(columns=["Unnamed:0"], inplace=True, errors="ignore")
```

```python
#temp
temperature.isnull().sum()
temperature["avg_temp"] = temperature["avg_temp"].fillna(
    temperature["avg_temp"].mean()
)
```

```python
fao.drop(columns=["Note"], inplace=True)
fao["Value"] = fao["Value"].fillna(fao["Value"].median())
```

```python
crop.to_csv("../processed/crop_recommendation_clean.csv", index=False)

weather.to_csv("../processed/weather_clean.csv", index=False)

soil.to_csv("../processed/soil_clean.csv", index=False)

yield_df.to_csv("../processed/yield_df_clean.csv", index=False)

yield_data.to_csv("../processed/yield_clean.csv", index=False)

rainfall.to_csv("../processed/rainfall_clean.csv", index=False)

temperature.to_csv("../processed/temperature_clean.csv", index=False)

pesticides.to_csv("../processed/pesticides_clean.csv", index=False)

fao.to_csv("../processed/faostat_clean.csv", index=False)
```
